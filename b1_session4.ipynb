```python
import pandas as pd

df = pd.read_csv("session4_dummy_students_data.csv")
print(df)
```

```python
print(df.head())
print(df.columns)
print(df.shape)
print(df.info())
```

```python
#convert math score column to numpy
math_scores =df["math"].to_numpy()
print(math_scores)
print(type(math_scores))
print(math_scores.mean())
```

```python
#calculate avg accross all subjects
df["average"] = df[["math", "science", "english"]].mean(axis=1)
print(df[["name", "math", "science", "english", "average"]])

```

```python
# 3 filters
#1st filter top students whose avg is greater than 85
#2nd low attenadance, attendance less than 75
#filter students from mathura

top_students =df[df["average"] >=85]
print(top_students[["name","average"]])


low_attendance = df[df["attendance_percent"] < 75]
print(low_attendance[["name", "attendance_percent"]])

mumbai_students = df[df["city"] == "Mumbai"]
print(mumbai_students[["name", "city", "average"]])
```
